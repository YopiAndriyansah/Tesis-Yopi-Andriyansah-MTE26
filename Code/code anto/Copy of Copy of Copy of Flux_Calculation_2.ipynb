```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Mempersiapkan Data

```python
# Daca data csv
df = pd.read_csv('/content/drive/MyDrive/Data Fluks/Hasil/Data_C-Flux_new_parameter')
df['timestamp'] = pd.to_datetime(df['timestamp'])
```

```python
# Tentukan interval
start_time = pd.to_datetime("2024-06-10 13:06:01.353000")
end_time   = pd.to_datetime("2024-06-10 13:36:01.028000")

# Filter data sesuai interval
df_interval = df[(df["timestamp"] >= start_time) & (df["timestamp"] <= end_time)]

num_rows = df_interval.shape[0]  # jumlah baris sebenarnya

print(f"Jumlah baris (record): {num_rows}")

# Jika ingin persentase kelengkapan
ideal_count = 7 * 60 * 30  # 30 menit * 7 Hz * 60 detik = 12600
accuracy_percent = (num_rows / ideal_count * 100)
print(f"Kelengkapan data: {accuracy_percent}%")


# Hitung rata-rata setiap kolom numerik
mean_values = df_interval.mean(numeric_only=True)

print("Rata-rata sensor pada interval:")
print(mean_values)
```
